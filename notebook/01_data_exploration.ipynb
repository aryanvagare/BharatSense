## 3. Hinglish dataset

**Reminder:** `tentative_sentiment` is a guess based on sampling, not confirmed.
Spot-check a sample of each label before trusting it as ground truth.

```python
hinglish = pd.read_csv('../data/interim/hinglish_interim.csv')
print('Rows:', len(hinglish))
hinglish.head()
```

```python
print('Numeric label distribution:')
print(hinglish['label'].value_counts())
print('\nTentative sentiment mapping (0=Negative, 1=Neutral, 2=Positive):')
print(hinglish['tentative_sentiment'].value_counts())
```

```python
# Manually spot-check samples from each label before trusting the mapping
for lbl in sorted(hinglish['label'].unique()):
    print(f'\n=== label={lbl} sample texts ===')
    for t in hinglish[hinglish['label'] == lbl]['text'].sample(5, random_state=1):
        print(' -', t[:120])
```
